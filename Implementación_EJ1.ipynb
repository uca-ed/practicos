```python
import json

def grafojson(archie):
    with open(archie, 'r') as f:
        estructura = json.load(f)
    return estructura['E']

def es_reflexivo(grafo):
    return all(nodo in vecinos for nodo, vecinos in grafo.items())

def es_simetrico(grafo):
    for nodo, vecinos in grafo.items():
        for vecino in vecinos:
            if nodo not in grafo.get(vecino, []):
                return False
    return True

def es_antisimetrico(grafo):
    for nodo, vecinos in grafo.items():
        for vecino in vecinos:
            if vecino != nodo and nodo in grafo.get(vecino, []):
                return False
    return True

def es_transitivo(grafo):
    for a, vecinos_a in grafo.items():
        for b in vecinos_a:
            vecinos_b = grafo.get(b, [])
            for c in vecinos_b:
                if c not in grafo[a]:
                    return False
    return True

grafo = grafojson("01.json")
print("Reflexivo:", es_reflexivo(grafo))
print("Simétrico:", es_simetrico(grafo))
print("Antisimétrico:", es_antisimetrico(grafo))
print("Transitivo:", es_transitivo(grafo))

if es_reflexivo(grafo) and es_simetrico(grafo) and es_transitivo(grafo):
    print("Es una relación de equivalencia")
elif es_reflexivo(grafo) and es_antisimetrico(grafo) and es_transitivo(grafo):
    print("Es un orden parcial")
else:
    print("No es ni orden ni equivalencia")
```

```python
import json
from collections import deque

def grafojson(exo):
    with open(exo, 'r') as f:
        estructura = json.load(f)
    return estructura["E"]

def camino1(grafo, inicio, destino):
    visitados = set()
    cola = deque([[inicio]])

    while cola:
        camino = cola.popleft()
        nodo_actual = camino[-1]

        if nodo_actual == destino:
            return camino

        if nodo_actual not in visitados:
            visitados.add(nodo_actual)
            for vecino in grafo.get(nodo_actual, []):
                nuevo_camino = list(camino)
                nuevo_camino.append(vecino)
                cola.append(nuevo_camino)

    return None  # Si no se encuentro camino

grafo = grafojson("esDivisorDe-200.json")

# Obtener primer y último nodo
nodos = list(grafo.keys())
inicio = nodos[0]
destino = nodos[-1]

print("Nodo inicial: ", inicio)
print("Nodo final: ", destino)

camino = camino1(grafo, inicio, destino)

if camino:
    print("Camino encontrado:", " a ".join(camino))
else:
    print("No existe camino entre ", inicio, "y", destino)
```
