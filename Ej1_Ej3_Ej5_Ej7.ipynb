```python
#Ej 1

class Cola:

	def __init__(self): #Iniciliazamos la variable
		self.elementos = []

	def vacio(self): #Verifica si la cola esta vacia
		if len(self.elementos) == 0:
			return True
		else:
			return False

	def insertar(self,nodo): #Encolar, instertar
		self.elementos.append(nodo)
		print("El elemento fue añadido")

	def eliminar(self): #Desencolar, eliminar
		if self.vacio():
			print("La estructura está vacía")
		else:
			eliminado = self.elementos.pop(0)
			print("Elemento eliminado : "+str(eliminado))

	def consultar(self):
		if self.vacio():
			print("La cola no tiene elementos")
		else:
			print(self.elementos)

obj = Cola()
while True: #Se muestra el menu de forma indefnida
	print("1.- Insertar (Encolar)")
	print("2.- Consultar")
	print("3.- Eliminar (Desencolar)")
	print("5.- Salir")
	opcion = int(input("Escribe tu opcion:"))
	if opcion==1:
		nodo = input("Escribe el nuevo elemento: ")
		obj.insertar(nodo)
	elif opcion==2:
		obj.consultar()
	elif opcion==3:
		obj.eliminar()
	elif opcion==4:
		break
	else:
		print("No existe esa opción")
```

```python
#Ej 5

from google.colab import files
uploaded = files.upload()
from collections import deque


class Graph:
    def __init__(self, edges, n):
        self.adjList = [[] for _ in range(n)]
        self.indegree = [0] * n
        for (src, dest) in edges:
            self.adjList[src].append(dest)
            self.indegree[dest] += 1


def doTopologicalSort(graph, n):
    L = []
    indegree = graph.indegree.copy()
    S = deque([i for i in range(n) if indegree[i] == 0])

    while S:
        node = S.pop()
        L.append(node)

        for m in graph.adjList[node]:
            indegree[m] -= 1
            if indegree[m] == 0:
                S.append(m)

    if any(indegree[i] > 0 for i in range(n)):
        return None

    return L


def leer_grafo_desde_archivo(nombre_archivo):
    edges = []
    max_nodo = -1
    with open(nombre_archivo, 'r') as f:
        for linea in f:
            src, dest = map(int, linea.strip().split())
            edges.append((src, dest))
            max_nodo = max(max_nodo, src, dest)
    return edges, max_nodo + 1


if __name__ == '__main__':
    archivo = 'grafo.txt'  # Cambia esto si tu archivo tiene otro nombre
    edges, n = leer_grafo_desde_archivo(archivo)
    graph = Graph(edges, n)
    orden = doTopologicalSort(graph, n)

    if orden:
        print("Orden topológico:", orden)
    else:
        print("El grafo tiene al menos un ciclo. No se puede hacer ordenamiento topológico.")
```
