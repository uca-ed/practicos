```python
##1## -- Cola -- 
## Archivo de entrada esperado: 
#                            ENQUEUE 5
#                            ENQUEUE 8
#                            DEQUEUE
#                            ENQUEUE 10
def operar_cola(nombre_archivo):
    cola = []

    with open(nombre_archivo, 'r') as archivo:
        for linea in archivo:
            partes = linea.strip().split()
            if not partes:
                continue
            operacion = partes[0].upper()

            if operacion == 'ENQUEUE' and len(partes) > 1:
                valor = partes[1]
                cola.append(valor)
            elif operacion == 'DEQUEUE':
                if cola:
                    cola.pop(0)
    
    print("Estado final de la cola:", cola)

# Ejemplo:
# operar_cola('operaciones_cola.txt')
```

```python
##2## -- Pila -- 
## Archivo de entrada esperado:
#                             PUSH 3
#                             PUSH 9
#                             POP
#                             PUSH 7

def operar_pila(nombre_archivo):
    pila = []

    with open(nombre_archivo, 'r') as archivo:
        for linea in archivo:
            partes = linea.strip().split()
            if not partes:
                continue
            operacion = partes[0].upper()

            if operacion == 'PUSH' and len(partes) > 1:
                valor = partes[1]
                pila.append(valor)
            elif operacion == 'POP':
                if pila:
                    pila.pop()
    
    print("Estado final de la pila:", pila)

# Ejemplo:
# operar_pila('operaciones_pila.txt')
```

```python
##3## -- Lista Enlazada -- 
def crear_nodo(dato):
    return {"dato": dato, "siguiente": None}

def insertar_al_final(lista, dato):
    nuevo = crear_nodo(dato)
    if lista is None:
        return nuevo
    actual = lista
    while actual["siguiente"] is not None:
        actual = actual["siguiente"]
    actual["siguiente"] = nuevo
    return lista

def eliminar(lista, valor):
    actual = lista
    anterior = None
    while actual is not None and actual["dato"] != valor:
        anterior = actual
        actual = actual["siguiente"]
    if actual is None:
        return lista  # No se encontró el valor
    if anterior is None:
        return actual["siguiente"]  # Se elimina el primer nodo
    else:
        anterior["siguiente"] = actual["siguiente"]
        return lista

def mostrar(lista):
    actual = lista
    while actual is not None:
        print(actual["dato"], end=" -> ")
        actual = actual["siguiente"]
    print("None")

#Ejemplo:
lista = None
lista = insertar_al_final(lista, 5)
lista = insertar_al_final(lista, 10)
lista = insertar_al_final(lista, 15)

mostrar(lista)  # 5 -> 10 -> 15 -> None

lista = eliminar(lista, 10)
mostrar(lista)  # 5 -> 15 -> None
```

```python
##4## -- Radix Sort -- 
## Archivo de entrada esperado (ejemplo):
# palabras.txt :
#              gato
#              perro
#              loro
#              pez
#              abeja
#              rana

def leer_palabras(nombre_archivo):
    with open(nombre_archivo, 'r') as archivo:
        return [linea.strip() for linea in archivo if linea.strip()]

def radix_sort_palabras(palabras):
    if not palabras:
        return []

    max_len = max(len(p) for p in palabras)
    
    for i in range(max_len - 1, -1, -1):  # desde el último carácter al primero
        buckets = [[] for _ in range(256)]  
        for palabra in palabras:
            char = ord(palabra[i]) if i < len(palabra) else 0 
            buckets[char].append(palabra)
        palabras = [palabra for bucket in buckets for palabra in bucket]

    return palabras

def main():
    archivo = "palabras.txt"
    palabras = leer_palabras(archivo)
    ordenadas = radix_sort_palabras(palabras)

    print("Palabras ordenadas:")
    for p in ordenadas:
        print(p)

# Ejecutar:
# main()
```

```python
##5## -- T-Sort basado en un grafo -- 
# Formato del archivo de entrada (grafo.txt) : 
# Cada línea representa una arista origen -> destino
#                                                   A B
#                                                   A C
#                                                   B D
#                                                   C D

from collections import defaultdict, deque

def leer_grafo(archivo):
    grafo = defaultdict(list)
    grados_entrada = defaultdict(int)

    with open(archivo, 'r') as f:
        for linea in f:
            if not linea.strip():
                continue
            origen, destino = linea.strip().split()
            grafo[origen].append(destino)
            grados_entrada[destino] += 1
            if origen not in grados_entrada:
                grados_entrada[origen] = 0

    return grafo, grados_entrada

def t_sort(grafo, grados_entrada):
    cola = deque([nodo for nodo in grados_entrada if grados_entrada[nodo] == 0])
    resultado = []

    while cola:
        actual = cola.popleft()
        resultado.append(actual)
        for vecino in grafo[actual]:
            grados_entrada[vecino] -= 1
            if grados_entrada[vecino] == 0:
                cola.append(vecino)

    if len(resultado) == len(grados_entrada):
        return resultado
    else:
        return None  # Hay un ciclo

def main():
    archivo = "grafo.txt"
    grafo, grados = leer_grafo(archivo)
    resultado = t_sort(grafo, grados)

    if resultado is None:
        print("La estructura es cíclica. No se puede realizar T-Sort.")
    else:
        print("Orden topológico (T-Sort):")
        print(" -> ".join(resultado))

# Ejecutar:
# main()
# Resultado esperado : A -> B -> C -> D
```

```python
##6##

import random
import time

#Creacion y carga de datos de las estructuras
def calcular_indice(edificio, piso, ala, aula, bloque):
    return (
        edificio * 5 * 2 * 25 * 17 +
        piso * 2 * 25 * 17 +
        ala * 25 * 17 +
        aula * 17 +
        bloque
    )

EDIFICIOS, PISOS, ALAS, AULAS, BLOQUES = 4, 5, 2, 25, 17
TOTAL = EDIFICIOS * PISOS * ALAS * AULAS * BLOQUES


INSCRIPTOS = [0] * TOTAL
CAPACIDAD = [random.randint(20, 60) for _ in range(TOTAL)]  # aula con capacidad entre 20 y 60

# Carga simulada (alumnos inscriptos menor o igual a la capacidad)
for i in range(TOTAL):
    INSCRIPTOS[i] = random.randint(0, CAPACIDAD[i])


# a) Aula/bloque horario con mayor porcentaje de ocupación
def aula_mayor_ocupacion():
    max_porcentaje = -1
    mejor_indice = -1
    for i in range(TOTAL):
        if CAPACIDAD[i] > 0:
            porcentaje = INSCRIPTOS[i] / CAPACIDAD[i]
            if porcentaje > max_porcentaje:
                max_porcentaje = porcentaje
                mejor_indice = i
    return mejor_indice, max_porcentaje

# b) Promedio de alumnos por piso en un bloque horario
def promedio_alumnos_por_piso(bloque_horario):
    promedios = []
    for piso in range(PISOS):
        suma = 0
        cuenta = 0
        for edificio in range(EDIFICIOS):
            for ala in range(ALAS):
                for aula in range(AULAS):
                    idx = calcular_indice(edificio, piso, ala, aula, bloque_horario)
                    suma += INSCRIPTOS[idx]
                    cuenta += 1
        promedios.append(suma / cuenta)
    return promedios

#c) Alumnos por ala dado edificio, piso y bloque horario
def alumnos_por_ala(edificio, piso, bloque_horario):
    resumen = [0, 0]  # [norte, sur]
    for ala in range(ALAS):
        total = 0
        for aula in range(AULAS):
            idx = calcular_indice(edificio, piso, ala, aula, bloque_horario)
            total += INSCRIPTOS[idx]
        resumen[ala] = total
    return resumen

#Comparar tiempos: acceso 5D vs 1D
def medir_tiempo_consulta(func, *args):
    inicio = time.time()
    resultado = func(*args)
    fin = time.time()
    return resultado, fin - inicio

# Prueba completa
def main():
    print("Aula con mayor ocupación:")
    idx, porc = aula_mayor_ocupacion()
    print(f"Índice: {idx}, Ocupación: {porc:.2%}")

    print("\nPromedio por piso para bloque horario 5:")
    print(promedio_alumnos_por_piso(5))

    print("\nAlumnos por ala (edificio 2, piso 3, bloque 8):")
    print(alumnos_por_ala(2, 3, 8))

    # Tiempos
    _, t1 = medir_tiempo_consulta(aula_mayor_ocupacion)
    _, t2 = medir_tiempo_consulta(promedio_alumnos_por_piso, 5)
    _, t3 = medir_tiempo_consulta(alumnos_por_ala, 2, 3, 8)

    print("\nTiempos de respuesta (en segundos):")
    print(f"a) Aula mayor ocupación: {t1:.6f}s")
    print(f"b) Promedio por piso: {t2:.6f}s")
    print(f"c) Alumnos por ala: {t3:.6f}s")

#Ejecutar:
# main()
```

```python
##7## -- Sort topologico sobre un grafo -- 
# Formato del archivo (grafo.txt):
#                                 A B
#                                 A C
#                                 B D
#                                 C D

from collections import defaultdict, deque

def leer_grafo(nombre_archivo):
    grafo = defaultdict(list)
    grados_entrada = defaultdict(int)

    with open(nombre_archivo, 'r') as archivo:
        for linea in archivo:
            if not linea.strip():
                continue
            origen, destino = linea.strip().split()
            grafo[origen].append(destino)
            grados_entrada[destino] += 1
            if origen not in grados_entrada:
                grados_entrada[origen] = 0  

    return grafo, grados_entrada

def topological_sort(grafo, grados_entrada):
    cola = deque([nodo for nodo in grados_entrada if grados_entrada[nodo] == 0])
    resultado = []

    while cola:
        nodo = cola.popleft()
        resultado.append(nodo)
        for vecino in grafo[nodo]:
            grados_entrada[vecino] -= 1
            if grados_entrada[vecino] == 0:
                cola.append(vecino)

    if len(resultado) == len(grados_entrada):
        return resultado
    else:
        return None  # Grafo cíclico

def main():
    grafo, grados = leer_grafo("grafo.txt")
    orden = topological_sort(grafo, grados)

    if orden is None:
        print("El grafo es cíclico. No se puede aplicar sort topológico.")
    else:
        print("Orden topológico:")
        print(" -> ".join(orden))

# Ejecutar:
# main()
```
