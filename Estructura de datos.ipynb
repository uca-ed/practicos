```python
# Minimales: nodos que no reciben conexiones (columnas vacías) = FUENTES
def minimales(matriz):
    return [j for j in range(matriz.shape[1]) if matriz.iloc[:, j].sum() == 0]

# Maximales: nodos que no emiten conexiones (filas vacías) = SUMIDEROS
def maximales(matriz):
    return [i for i in range(matriz.shape[0]) if matriz.iloc[i, :].sum() == 0]

# Vecindad derecha: nodos a los que apunta un nodo
def vecindad_derecha(matriz, nodo):
    return [j for j in range(matriz.shape[1]) if matriz.iat[nodo, j] == 1]

# Vecindad izquierda: nodos que apuntan al nodo
def vecindad_izquierda(matriz, nodo):
    return [i for i in range(matriz.shape[0]) if matriz.iat[i, nodo] == 1]
```

```python
for idx, matriz in enumerate(matrices):
    print(f'\nMatriz {idx}:')
    print(f'  Maximales: {maximales(matriz)}')
    print(f'  Minimales: {minimales(matriz)}')

    for nodo in range(matriz.shape[0]):
        print(f'    Nodo {nodo} -> VD: {vecindad_derecha(matriz, nodo)} | VI: {vecindad_izquierda(matriz, nodo)}')
```

### Ejercicio 2
Verificar si un grafo dado cumple las propiedades de Reflexividad, Simetría, Antisimetría y Transitividad. Luego, debe decidir si el grafo es un orden o es corresponde a una relación de equivalencia

```python
def cargar_json(path):
    with open(path, 'r') as f:
        data = json.load(f)
    nodos = list(map(int, data["P"]))
    adyacencia = {int(k): list(map(int, v)) for k, v in data["E"].items()}
    return nodos, adyacencia
```

```python
nodos_1, adyacencias_1 = cargar_json('/content/drive/MyDrive/Estructura de datis/Archivos_2/01.json')
nodos_2, adyacencias_2 = cargar_json('/content/drive/MyDrive/Estructura de datis/Archivos_2/02.json')
nodos_3, adyacencias_3 = cargar_json('/content/drive/MyDrive/Estructura de datis/Archivos_2/03.json')
```

```python
# Para todo n tiene que existir un par (n,n)
def es_reflexivo(nodos, ady):
    return all(n in ady.get(n, []) for n in nodos)

# Si (a,b) entonces (b, a)
def es_simetrico(ady):
    for a, vecinos in ady.items():
        for b in vecinos:
            if a not in ady.get(b, []):
                return False
    return True

# Si (a,b) y (b,a) enontces no siempre a == b
def es_antisimetrico(ady):
    for a, vecinos in ady.items():
        for b in vecinos:
            if b != a and a in ady.get(b, []):
                return False
    return True


# Si (a,b) y (b,c) entonces (a,c)
def es_transitivo(ady):
    for a, vecinos in ady.items():
        for b in vecinos:
            for c in ady.get(b, []):
                if c not in ady.get(a, []):
                    return False
    return True

# Aca simplemente imprimo y veo si son trues para calcular equivalencia u orden
def analizar_grafo(nodos, ady):
    reflexivo = es_reflexivo(nodos, ady)
    simetrico = es_simetrico(ady)
    antisimetrico = es_antisimetrico(ady)
    transitivo = es_transitivo(ady)

    if reflexivo and simetrico and transitivo:
        print("Es una RELACIÓN DE EQUIVALENCIA")
    elif reflexivo and antisimetrico and transitivo:
        print("Es una RELACIÓN DE ORDEN")
    else:
        print("No es ni una relación de equivalencia ni de orden")

    print(f"Reflexivo: {reflexivo}")
    print(f"Simétrico: {simetrico}")
    print(f"Antisimétrico: {antisimetrico}")
    print(f"Transitivo: {transitivo}")
```

```python
analizar_grafo(nodos_2, adyacencias_2)
```
