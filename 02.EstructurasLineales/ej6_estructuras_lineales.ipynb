```python
import numpy as np
import random
import time
import pandas as pd
import matplotlib.pyplot as plt

# Definir las dimensiones y tamaños
edificio_size = 4
piso_size = 5
ala_size = 2
aula_size = 25
bloque_horario_size = 17

# Calcula el tamaño total del arreglo unidimensional
total_size = edificio_size * piso_size * ala_size * aula_size * bloque_horario_size

print(f"Tamaño total necesario para los arreglos unidimensionales: {total_size}")

# Crear funciones para mapear índices
def map_5d_to_1d(edificio, piso, ala, aula, bloque_horario):
  """
  Mapea índices 5D (edificio, piso, ala, aula, bloque_horario) a un único índice 1D.

  Args:
    edificio: Índice para la dimensión del edificio (basado en 0).
    piso: Índice para la dimensión del piso (basado en 0).
    ala: Índice para la dimensión del ala (basado en 0).
    aula: Índice para la dimensión del aula (basado en 0).
    bloque_horario: Índice para la dimensión del bloque_horario (basado en 0).

  Returns:
    El índice 1D correspondiente.
  """
  index_1d = edificio * (piso_size * ala_size * aula_size * bloque_horario_size) + \
             piso * (ala_size * aula_size * bloque_horario_size) + \
             ala * (aula_size * bloque_horario_size) + \
             aula * (bloque_horario_size) + \
             bloque_horario
  return index_1d

# Crear arreglos unidimensionales
INSCRIPTOS = np.zeros(total_size, dtype=int)
CAPACIDAD = np.zeros(total_size, dtype=int)

print("\nArreglo INSCRIPTOS inicializado con ceros:")
display(INSCRIPTOS[:10])
print("\nArreglo CAPACIDAD inicializado con ceros:")
display(CAPACIDAD[:10])

# Cargar datos de ejemplo
print("\nCargando datos de ejemplo...")
for edificio in range(edificio_size):
  for piso in range(piso_size):
    for ala in range(ala_size):
      for aula in range(aula_size):
        for bloque_horario in range(bloque_horario_size):
          capacidad = random.randint(20, 100)
          inscriptos = random.randint(0, capacidad)
          index_1d = map_5d_to_1d(edificio, piso, ala, aula, bloque_horario)
          CAPACIDAD[index_1d] = capacidad
          INSCRIPTOS[index_1d] = inscriptos

print("Carga de datos completada.")
print("\nArreglo INSCRIPTOS poblado (primeros 20 elementos):")
display(INSCRIPTOS[:20])
print("\nArreglo CAPACIDAD poblado (primeros 20 elementos):")
display(CAPACIDAD[:20])

# Implementar algoritmo para el interrogante a (Mayor porcentaje de ocupación)
print("\nEjecutando algoritmo para mayor porcentaje de ocupación...")
max_occupation_percentage = -1
best_edificio = -1
best_piso = -1
best_ala = -1
best_aula = -1
best_bloque_horario = -1

start_time_a = time.time()
for edificio in range(edificio_size):
    for piso in range(piso_size):
        for ala in range(ala_size):
            for aula in range(aula_size):
                for bloque_horario in range(bloque_horario_size):
                    index_1d = map_5d_to_1d(edificio, piso, ala, aula, bloque_horario)
                    inscriptos = INSCRIPTOS[index_1d]
                    capacidad = CAPACIDAD[index_1d]
                    if capacidad > 0:
                        occupation_percentage = (inscriptos / capacidad) * 100
                        if occupation_percentage > max_occupation_percentage:
                            max_occupation_percentage = occupation_percentage
                            best_edificio = edificio
                            best_piso = piso
                            best_ala = ala
                            best_aula = aula
                            best_bloque_horario = bloque_horario
end_time_a = time.time()
time_a = end_time_a - start_time_a

print(f"Aula/bloque horario con mayor porcentaje de ocupación:")
print(f"  Edificio: {best_edificio}")
print(f"  Piso: {best_piso}")
print(f"  Ala: {best_ala}")
print(f"  Aula: {best_aula}")
print(f"  Bloque horario: {best_bloque_horario}")
print(f"  Porcentaje de ocupación: {max_occupation_percentage:.2f}%")

# Implementar algoritmo para el interrogante b (Promedio de alumnos por piso en un bloque horario)
print("\nEjecutando algoritmo para promedio de alumnos por piso/bloque horario...")
def average_students_by_piso_bloque_horario(piso_input, bloque_horario_input):
  total_students = 0
  classroom_count = 0
  for edificio in range(edificio_size):
    for ala in range(ala_size):
      for aula in range(aula_size):
        index_1d = map_5d_to_1d(edificio, piso_input, ala, aula, bloque_horario_input)
        total_students += INSCRIPTOS[index_1d]
        classroom_count += 1
  if classroom_count > 0:
    average_students = total_students / classroom_count
  else:
    average_students = 0
  return average_students

# Ejemplo de uso y medición de tiempo
example_piso_b = 2
example_bloque_horario_b = 5

start_time_b = time.time()
average_b = average_students_by_piso_bloque_horario(example_piso_b, example_bloque_horario_b)
end_time_b = time.time()
time_b = end_time_b - start_time_b

print(f"Promedio de alumnos para el piso {example_piso_b} en el bloque horario {example_bloque_horario_b}: {average_b:.2f}")

# Implementar algoritmo para el interrogante c (Cantidad total de alumnos por ala)
print("\nEjecutando algoritmo para cantidad total de alumnos por ala...")
def total_students_by_ala_for_location(edificio_input, piso_input, bloque_horario_input):
  students_by_ala = {0: 0, 1: 0}
  for ala in range(ala_size):
    for aula in range(aula_size):
      index_1d = map_5d_to_1d(edificio_input, piso_input, ala, aula, bloque_horario_input)
      students_by_ala[ala] += INSCRIPTOS[index_1d]
  return students_by_ala

# Ejemplo de uso y medición de tiempo
example_edificio_c = 1
example_piso_c = 3
example_bloque_horario_c = 7

start_time_c = time.time()
total_students_alas_c = total_students_by_ala_for_location(example_edificio_c, example_piso_c, example_bloque_horario_c)
end_time_c = time.time()
time_c = end_time_c - start_time_c

print(f"Cantidad total de alumnos para Edificio {example_edificio_c}, Piso {example_piso_c}, Bloque Horario {example_bloque_horario_c}:")
print(f"  Ala Norte (0): {total_students_alas_c[0]}")
print(f"  Ala Sur (1): {total_students_alas_c[1]}")


# Realizar pruebas y medir tiempos (Medición ya incluida en pasos 5, 6 y 7) y Presentar resultados
print("\n--- Informe de Tiempos y Resultados ---")

# Crear un DataFrame para almacenar los tiempos de ejecución
data_times = {'Algoritmo': ['Mayor Porcentaje de Ocupación', 'Promedio de Alumnos por Piso/Bloque Horario', 'Cantidad Total de Alumnos por Ala'],
              'Tiempo de Ejecución (s)': [time_a, time_b, time_c]}
df_times = pd.DataFrame(data_times)

print("Tiempos de Ejecución:")
display(df_times)

# Mostrar resultados de los algoritmos
print("\nResultados de los Algoritmos (con valores de ejemplo):")
print(f"a. Aula/bloque horario con mayor porcentaje de ocupación: Edificio={best_edificio}, Piso={best_piso}, Ala={best_ala}, Aula={best_aula}, Bloque Horario={best_bloque_horario}, Porcentaje de Ocupación={max_occupation_percentage:.2f}%")
print(f"b. Promedio de alumnos para el Piso {example_piso_b} en el Bloque Horario {example_bloque_horario_b}: {average_b:.2f}")
print(f"c. Cantidad total de alumnos para Edificio {example_edificio_c}, Piso {example_piso_c}, Bloque Horario {example_bloque_horario_c}: Ala Norte (0): {total_students_alas_c[0]}, Ala Sur (1): {total_students_alas_c[1]}")

# Crear y mostrar un gráfico de barras para visualizar los tiempos de ejecución
plt.figure(figsize=(12, 7))
plt.bar(df_times['Algoritmo'], df_times['Tiempo de Ejecución (s)'], color=['blue', 'green', 'red'])
plt.title('Tiempo de Ejecución por Algoritmo')
plt.xlabel('Algoritmo')
plt.ylabel('Tiempo de Ejecución (s)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```
