```python
def apilar(pila, valor):
    pila.append(valor)

def desapilar(pila):
    if not esta_vacia(pila):
        return pila.pop()
    else:
        return None

def esta_vacia(pila):
    return len(pila) == 0

# Inicializar una pila vacía
pila = []

# Leer operaciones del archivo
try:
    with open('operations_test.txt', 'r') as f:
        for linea in f:
            operacion = linea.strip().split()
            if operacion[0].upper() == 'PUSH':
                try:
                    valor = int(operacion[1])
                    apilar(pila, valor)
                except IndexError:
                    print("Error: La operación PUSH requiere un valor.")
                except ValueError:
                    print("Error: Valor inválido para la operación PUSH.")
            elif operacion[0].upper() == 'POP':
                valor_desapilado = desapilar(pila)
                if valor_desapilado is not None:
                    print(f"Desapilado: {valor_desapilado}")
                else:
                    print("Error: No se puede desapilar de una pila vacía.")
            else:
                print(f"Operación desconocida: {operacion[0]}")
except FileNotFoundError:
    print("Error: operations_test.txt no encontrado.")

# Mostrar la pila final
print("Pila final:", pila)
```
