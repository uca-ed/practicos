```python
class Celda:
    """Clase que representa una celda en una lista enlazada simple"""
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None  # Enlace a la siguiente celda

class ListaEnlazada:
    """Clase que representa una lista enlazada simple"""
    def __init__(self):
        self.primera = None  # Referencia a la primera celda
        self.longitud = 0
    
    def esta_vacia(self):
        """Devuelve True si la lista está vacía"""
        return self.primera is None
    
    def agregar(self, valor):
        """Agrega un elemento al final de la lista"""
        nueva_celda = Celda(valor)
        if self.esta_vacia():
            self.primera = nueva_celda
        else:
            actual = self.primera
            while actual.siguiente is not None:
                actual = actual.siguiente
            actual.siguiente = nueva_celda
        self.longitud += 1
    
    def insertar(self, indice, valor):
        """Inserta un elemento en la posición especificada"""
        if indice < 0 or indice > self.longitud:
            raise IndexError("Índice fuera de rango")
        
        nueva_celda = Celda(valor)
        
        if indice == 0:
            nueva_celda.siguiente = self.primera
            self.primera = nueva_celda
        else:
            actual = self.primera
            for _ in range(indice - 1):
                actual = actual.siguiente
            nueva_celda.siguiente = actual.siguiente
            actual.siguiente = nueva_celda
        
        self.longitud += 1
    
    def eliminar(self, valor):
        """Elimina la primera ocurrencia del valor especificado"""
        if self.esta_vacia():
            raise ValueError("La lista está vacía")
        
        if self.primera.valor == valor:
            self.primera = self.primera.siguiente
            self.longitud -= 1
            return
        
        actual = self.primera
        while actual.siguiente is not None and actual.siguiente.valor != valor:
            actual = actual.siguiente
        
        if actual.siguiente is None:
            raise ValueError("El valor no está en la lista")
        
        actual.siguiente = actual.siguiente.siguiente
        self.longitud -= 1
    
    def __len__(self):
        """Devuelve la longitud de la lista"""
        return self.longitud
    
    def __str__(self):
        """Representación en cadena de la lista"""
        elementos = []
        actual = self.primera
        while actual is not None:
            elementos.append(str(actual.valor))
            actual = actual.siguiente
        return " -> ".join(elementos) + " -> None"

# Ejemplo de uso
if __name__ == "__main__":
    lista = ListaEnlazada()
    print("Lista vacía:", lista)
    
    lista.agregar(10)
    lista.agregar(20)
    lista.agregar(30)
    print("Lista después de agregar elementos:", lista)
    
    lista.insertar(1, 15)
    print("Lista después de insertar 15 en posición 1:", lista)
    
    lista.eliminar(20)
    print("Lista después de eliminar el 20:", lista)
    
    print("Longitud de la lista:", len(lista))
```
