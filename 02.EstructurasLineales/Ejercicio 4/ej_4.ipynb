```python
def counting_sort(words, index, max_len):
    alphabet = ' abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ'
    count = [0] * (len(alphabet) + 1)
    output = [None] * len(words)
    
    for word in words:
        if index < len(word):
            char = word[len(word) - 1 - index]
            pos = alphabet.find(char) + 1 if char in alphabet else 0
        else:
            pos = 0
        count[pos] += 1
    
    for i in range(1, len(count)):
        count[i] += count[i-1]

    for word in reversed(words):
        if index < len(word):
            char = word[len(word) - 1 - index]
            pos = alphabet.find(char) + 1 if char in alphabet else 0
        else:
            pos = 0
        output[count[pos] - 1] = word
        count[pos] -= 1
    
    return output

def radix_sort_words(words):
    if not words:
        return []

    max_len = max(len(word) for word in words)
    
    for i in range(max_len):
        words = counting_sort(words, i, max_len)
    
    return words

def read_words_from_file(filename):
    with open(filename, 'r', encoding='utf-8') as file:
        text = file.read()
        words = text.split()
    return words

def write_words_to_file(filename, words):
    with open(filename, 'w', encoding='utf-8') as file:
        file.write(' '.join(words))

def sort_file(input_filename, output_filename):
    words = read_words_from_file(input_filename)
    sorted_words = radix_sort_words(words)
    write_words_to_file(output_filename, sorted_words)
    print(f"Palabras ordenadas guardadas en {output_filename}")

if __name__ == "__main__":
    input_file = input("Ingrese el nombre del archivo de entrada: ")
    output_file = input("Ingrese el nombre del archivo de salida: ")
    
    try:
        sort_file(input_file, output_file)
    except FileNotFoundError:
        print(f"Error: No se pudo encontrar el archivo {input_file}")
    except Exception as e:
        print(f"Ocurrió un error: {str(e)}")
```
