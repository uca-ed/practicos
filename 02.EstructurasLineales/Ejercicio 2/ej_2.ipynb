```python
class Pila:
    def __init__(self):
        self.items = []
    
    def apilar(self, item):
        self.items.append(item)
    
    def desapilar(self):
        if not self.esta_vacia():
            return self.items.pop()
        return None
    
    def esta_vacia(self):
        return len(self.items) == 0
    
    def cima(self):
        if not self.esta_vacia():
            return self.items[-1]
        return None
    
    def tamano(self):
        return len(self.items)
    
    def __str__(self):
        return str(self.items)

def procesar_operaciones(archivo):
    pila = Pila()
    
    with open(archivo, 'r') as f:
        for linea in f:
            linea = linea.strip()
            if not linea:
                continue
            
            if linea.startswith('apilar'):
                _, valor = linea.split()
                pila.apilar(valor)
                print(f"Apilando {valor}: {pila}")
            elif linea == 'desapilar':
                valor = pila.desapilar()
                print(f"Desapilando: {valor} - Pila actual: {pila}")
            elif linea == 'cima':
                print(f"Cima de la pila: {pila.cima()}")
            elif linea == 'vacía':
                print(f"¿La pila está vacía? {pila.esta_vacia()}")
            elif linea == 'tamaño':
                print(f"Tamaño de la pila: {pila.tamano()}")
            else:
                print(f"Operación no reconocida: {linea}")
    
    print("\nResultado final de la pila:")
    print(pila)

with open('operaciones_pila.txt', 'w') as f:
    f.write("apilar 10\n")
    f.write("apilar 20\n")
    f.write("apilar 30\n")
    f.write("cima\n")
    f.write("desapilar\n")
    f.write("vacía\n")
    f.write("tamaño\n")
    f.write("desapilar\n")
    f.write("desapilar\n")
    f.write("vacía\n")

procesar_operaciones('operaciones_pila.txt')
```
