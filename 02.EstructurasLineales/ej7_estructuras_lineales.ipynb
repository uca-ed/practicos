```python
from collections import defaultdict, deque

def ordenamiento_topologico(grafo):
    """
    Realiza un ordenamiento topológico en un grafo dirigido acíclico (DAG).
grafo: Un diccionario que representa el grafo donde las claves son nodos
               y los valores son listas de nodos adyacentes.

returns: Una lista que representa el orden topológico, o None si el grafo contiene un ciclo.
    """
    grado_entrada = {nodo: 0 for nodo in grafo}
    for nodo in grafo:
        for vecino in grafo[nodo]:
            grado_entrada[vecino] += 1

    cola = deque([nodo for nodo, grado in grado_entrada.items() if grado == 0])
    resultado = []

    while cola:
        u = cola.popleft()
        resultado.append(u)

        if u in grafo: # Comprobar si el nodo u tiene aristas salientes
            for v in grafo[u]:
                grado_entrada[v] -= 1
                if grado_entrada[v] == 0:
                    cola.append(v)

    if len(resultado) == len(grafo):
        return resultado
    else:
        return None

archivo_grafo = 'graph.txt'
grafo = defaultdict(list)
todos_los_nodos = set()

try:
    with open(archivo_grafo, 'r') as f:
        for linea in f:
            u, v = linea.strip().split()
            grafo[u].append(v)
            todos_los_nodos.add(u)
            todos_los_nodos.add(v)

    for nodo in todos_los_nodos:
        if nodo not in grafo:
            grafo[nodo] = []

    nodos_ordenados = ordenamiento_topologico(grafo)

    if nodos_ordenados:
        print("Orden de Ordenamiento Topológico:", nodos_ordenados)
    else:
        print("El grafo contiene un ciclo. El ordenamiento topológico no es posible.")

except FileNotFoundError:
    print(f"Error: Archivo '{archivo_grafo}' no encontrado.")
except Exception as e:
    print(f"Ocurrió un error: {e}")
```
