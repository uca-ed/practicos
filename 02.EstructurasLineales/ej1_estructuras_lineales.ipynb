```python
# Paso 1: Crear un archivo de operaciones de ejemplo
with open("operaciones_cola.txt", "w") as f:
    f.write("enqueue 10\n")
    f.write("enqueue 20\n")
    f.write("dequeue\n")
    f.write("enqueue 30\n")
    f.write("dequeue\n")
    f.write("enqueue 40\n")
    f.write("enqueue 50\n")
    f.write("dequeue\n")
```

```python
# Paso 2: Implementar la cola usando una lista (arreglo)
class Cola:
    def __init__(self):
        self.items = [] #lista de Python como nuestro "arreglo"

    def esta_vacia(self):
        return self.items == []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if not self.esta_vacia():
            return self.items.pop(0)
        else:
            return None

    def tamano(self):
        return len(self.items)

    def __str__(self):
        return str(self.items)

# Paso 3 y 4: Leer el archivo y procesar las operaciones
mi_cola = Cola() # cola vacía

try:
    with open("operaciones_cola.txt", "r") as f:
        for linea in f:
            linea = linea.strip().split()
            operacion = linea[0]

            if operacion == "enqueue":
                if len(linea) > 1:
                    try:
                        valor = int(linea[1])
                        mi_cola.enqueue(valor)
                        print(f"Operación: enqueue {valor} -> Cola: {mi_cola}")
                    except ValueError:
                        print(f"Error: Valor inválido para enqueue en la línea: {' '.join(linea)}")
                else:
                    print(f"Error: Falta valor para enqueue en la línea: {' '.join(linea)}")
            elif operacion == "dequeue":
                item_removido = mi_cola.dequeue()
                if item_removido is not None:
                    print(f"Operación: dequeue -> Elemento removido: {item_removido} -> Cola: {mi_cola}")
                else:
                    print(f"Operación: dequeue -> Cola vacía, no se pudo hacer dequeue -> Cola: {mi_cola}")
            else:
                print(f"Operación desconocida: {operacion} en la línea: {' '.join(linea)}")

except FileNotFoundError:
    print("Error: El archivo 'operaciones_cola.txt' no fue encontrado.")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")


# Paso 5: Mostrar el resultado final
print("\n--- Resultado Final ---")
print(f"Contenido final de la cola: {mi_cola}")
print(f"Tamaño final de la cola: {mi_cola.tamano()}")
```
