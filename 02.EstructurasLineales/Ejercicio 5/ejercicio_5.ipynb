```python
from collections import deque, defaultdict

class Graph:
    def __init__(self):
        # Lista de adyacencia: nodo -> lista de sucesores
        self.adj = defaultdict(list)
        
        # Grado de entrada de cada nodo
        self.in_degree = defaultdict(int)

    def add_edge(self, u, v):
        """Añade la arista dirigida u -> v."""
        self.adj[u].append(v)
        
        # Aseguramos que ambos nodos existan en in_degree
        self.in_degree[v] += 1
        self.in_degree.setdefault(u, 0)

    def topological_sort(self):
        """Devuelve una lista con un orden topogico si existe,
        o lanza ValueError si el grafo es cíclico."""
        
        # Cola de nodos con grado de entrada 0 (usaria la cola que hicimos anteriormente pero no se como jeje)
        queue = deque([u for u in self.in_degree if self.in_degree[u] == 0])
        topo_order = []

        # Procesar nodos sin predecesores
        while queue:
            u = queue.popleft()
            topo_order.append(u)

            for v in self.adj[u]:
                self.in_degree[v] -= 1
                if self.in_degree[v] == 0:
                    queue.append(v)

        if len(topo_order) != len(self.in_degree):
            raise ValueError("El grafo es ciclico")

        return topo_order


if __name__ == "__main__":
    # Testeo:

    g = Graph()
    # Añadimos aristas: A -> C, B -> C, C -> D, D -> E  
    g.add_edge('A', 'C')
    g.add_edge('B', 'C')
    g.add_edge('C', 'D')
    g.add_edge('D', 'E')
    # g.add_edge('X', 'Y')  # Aca por si queres crear un nuevo ciclo, pero lo testeamos con todo y funca

    try:
        order = g.topological_sort()
        print("Orden topologico:", order)
    except ValueError as e:
        print("Error:", e)
```
