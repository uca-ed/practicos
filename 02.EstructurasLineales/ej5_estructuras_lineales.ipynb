```python
from collections import defaultdict

class Graph:
    def __init__(self, vertices):
        # lista de adyacencia
        self.graph = defaultdict(list)
        self.V = vertices

    # Función para añadir una arista al grafo
    def add_edge(self, u, v):
        self.graph[u].append(v)

# Implementación del algoritmo T-Sort
def t_sort(graph):
    # Array para almacenar el grado de entrada de cada vértice
    in_degree = [0] * graph.V
    # Calcular el grado de entrada de cada vértice
    for i in range(graph.V):
        for j in graph.graph[i]:
            in_degree[j] += 1

    # Cola para almacenar los vértices con grado de entrada 0
    queue = []
    for i in range(graph.V):
        if in_degree[i] == 0:
            queue.append(i)

    # Lista para almacenar el resultado del T-Sort
    t_sorted_list = []
    # Contador de vértices visitados
    count = 0

    # Procesar los vértices en la cola
    while queue:
        # Obtener el primer vértice de la cola
        u = queue.pop(0)
        # Añadir el vértice a la lista de ordenamiento
        t_sorted_list.append(u)

        # Reducir el grado de entrada de los vértices adyacentes
        for v in graph.graph[u]:
            in_degree[v] -= 1
            # Si el grado de entrada se vuelve 0, añadir el vértice a la cola
            if in_degree[v] == 0:
                queue.append(v)
        # Incrementar el contador de vértices visitados
        count += 1

    # Comprobar si hay un ciclo en el grafo
    if count == graph.V:
        # Si el número de vértices visitados es igual al total de vértices, el grafo es acíclico
        return t_sorted_list
    else:
        # Si no, el grafo es cíclico
        return "El grafo es cíclico"

# Ejemplo de uso:
# Crear un grafo
g = Graph(6)
g.add_edge(5, 2)
g.add_edge(5, 0)
g.add_edge(4, 0)
g.add_edge(4, 1)
g.add_edge(2, 3)
g.add_edge(3, 1)

# Realizar el T-Sort
result = t_sort(g)
print(f"Resultado del T-Sort: {result}")

# Ejemplo con un grafo cíclico
g_cyclic = Graph(3)
g_cyclic.add_edge(0, 1)
g_cyclic.add_edge(1, 2)
g_cyclic.add_edge(2, 0)

result_cyclic = t_sort(g_cyclic)
print(f"Resultado del T-Sort para grafo cíclico: {result_cyclic}")
```
