```python
import glob
import os

# Función para crear archivos de ejemplo para demostración
def create_dummy_files():
    if not os.path.exists('/tmp/file1.txt'):
        with open('/tmp/file1.txt', 'w') as f:
            f.write("manzana banana cereza cita\n")
            f.write("uva fuego limon naranja\n")
    if not os.path.exists('/tmp/file2.txt'):
        with open('/tmp/file2.txt', 'w') as f:
            f.write("kiwi mango mandarina pera\n")
            f.write("quince frambuesa frutilla\n")

create_dummy_files()

# Función para cargar palabras de archivos
def load_words_from_files(file_patterns):
    words = []
    for pattern in file_patterns:
        for filename in glob.glob(pattern):
            with open(filename, 'r') as f:
                for line in f:
                    words.extend(''.join(filter(str.isalpha, word)).lower() for word in line.split())
    return words

# Función para realizar counting sort basado en el carácter en la posición
def counting_sort(arr, exp):
    n = len(arr)
    output = [0] * n
    count = [0] * 257  # Rango ASCII extendido (0-255) más uno para manejar cadenas más cortas

    for i in range(n):
        # Obtener el código del carácter en la posición actual (o 0 si la cadena es más corta)
        char_code = ord(arr[i][exp]) + 1 if exp < len(arr[i]) else 0
        count[char_code] += 1

    for i in range(1, 257):
        count[i] += count[i - 1]

    for i in range(n - 1, -1, -1):
        char_code = ord(arr[i][exp]) + 1 if exp < len(arr[i]) else 0
        output[count[char_code] - 1] = arr[i]
        count[char_code] -= 1

    for i in range(n):
        arr[i] = output[i]

# Función Radix Sort
def radix_sort(arr):
    # Encontrar la longitud máxima de una cadena en el array
    max_len = max(len(s) for s in arr)

    # Realizar counting sort para cada posición de carácter
    for exp in range(max_len - 1, -1, -1): # Iterar desde el último carácter hasta el primero
        counting_sort(arr, exp)

# Reemplaza con tus patrones de archivo
file_patterns = ['/tmp/file1.txt', '/tmp/file2.txt']

# Cargar las palabras
words_list = load_words_from_files(file_patterns)
print(f"Se cargaron {len(words_list)} palabras.")

# Ordenar las palabras usando Radix Sort
radix_sort(words_list)

# Mostrar las palabras ordenadas
print("\nPalabras ordenadas:")
print(words_list)
```
