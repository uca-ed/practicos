```python
class Celda:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None

class ListaEnlazada:
    def __init__(self):
        self.cabeza = None

    def agregar(self, valor):
        nueva_celda = Celda(valor)
        if self.cabeza is None:
            self.cabeza = nueva_celda
        else:
            actual = self.cabeza
            while actual.siguiente is not None:
                actual = actual.siguiente
            actual.siguiente = nueva_celda

    def imprimir(self):
        actual = self.cabeza
        while actual is not None:
            print(actual.valor, end=" -> ")
            actual = actual.siguiente
        print("None")

# Ejemplo de uso
mi_lista = ListaEnlazada()
mi_lista.agregar(10)
mi_lista.agregar(20)
mi_lista.agregar(30)
mi_lista.imprimir()
```
